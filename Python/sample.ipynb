Step 10 - Outlimiters

```python
q1 = df['Age'].quantile(0.25)
```

```python
q3 = df['Age'].quantile(0.75)
```

```python
IQR = q3-q1
```

```python
lower = q1-1.5*IQR
```

```python
upper = q3+1.5*IQR
```

```python
outliers = df[(df['Age']<lower) | (df['Age']>upper)]
outliers
```
