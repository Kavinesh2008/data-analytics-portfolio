```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("/content/drive/MyDrive/DataSets/Sample - Superstore.xls")
print(df.head())
```

```python
print(df.tail())
```

```python
#To check if we have any duplicate values
records = df.duplicated()
count = df.duplicated().count()
sum = df.duplicated().sum()
print(records)
print()
print(count)
print()
print(sum)
```

```python
#Univariate
#Bar,Pie,Count,Histogram,Box plot,Donut
plt.bar(df['Category'],df['Sales'],color=['red','yellow','blue','orange','green'],hatch='-',edgecolor='black')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.title('Bar Chart')
plt.show()
```

```python
#Histogram
plt.hist(df['Quantity'],color='green',hatch='x',edgecolor='black')
plt.xlabel('Quantity')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```

```python
plt.barh(df['Category'],df['Sales'],color='blue')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.title('Column Chart')
plt.show()
```

```python
#Bivariate
#Scatter,Line,
```

```python
plt.scatter(df['Profit'],df['Sales'],color='red',hatch='o',edgecolor='blue')
plt.xlabel('Profit')
plt.ylabel('Sales')
plt.title('Scatter Plot')
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure()
ax = fig.add_subplot(111,projection='3d')
ax.scatter(df['Profit'],df['Sales'],df['Quantity'])
plt.xlabel('Profit')
plt.ylabel('Sales')
plt.title('3D Scatter Plot')
plt.show()
```

```python
#Multivirate
#Heap Map,Cluster,Radar,Tree map
```

```python
#Pair plot
sns.pairplot(df,vars=['Sales','Quantity','Profit'],hue='Category')
plt.show()
```

```python
#Heat map
df1 = df[['Sales','Quantity','Profit']]
corr = df1.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap='viridis`')
plt.show()
```
